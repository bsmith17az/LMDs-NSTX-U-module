# 001: Evaporation rate calculation
## Purpose
Calculate the approximate evaporation rate of lithium from the surface of the module for comparison against simulation output and to estimate the resultant free-surface deformation
## Assumptions
- Uniform tile surface temperature
- No redeposition of atoms that have left the surface
- Hertz-Knudsen relation applies (implication?) (evaporation/langmuir limit with accommodation coefficient 1)
## Inputs
$p_{v,Li}[Pa]=10^{10.061-\frac{8023}{T_{surface}[Pa]}}$  
$\Gamma_{vap}=\frac{p_{v}}{\sqrt{2\pi m k_B T_{surface}}}$  - particle flux in $m^{-2}s^{-1}$  
(as used in Horacek2020)  
$\dot{m}''_{vap}=\frac{p_{v}}{\sqrt{\frac{m}{2\pi k_B T_{surface}}}}$  - mass flux in $kgm^{-2}s^{-1}$ 

$T_{surface} = 800K$  
$m_{Li} = 1.15 \times 10^{-26}kg$  
$k_B = 1.380649 \times 10^{-23}JK^{-1}$

```python
from math import sqrt, pi

T=[973, 1173] # lower bound, upper bound; Kelvin
t=17 # shot duration in s
m = 1.15 * 10 ** (-26) # mass of lithium atom in kg
kb = 1.38 * 10 ** (-23) # Boltzmann constant in J/K
A=0.0373*0.0397 # tile area in m^2

pv=[0, 0] # vapor pressure in Pa
gamma=[0, 0] # evaporation rate in m^-2s^-1
mdot=[0, 0] # mass flux in kg/m^2/s
m_loss=[0, 0] # mass loss over entire shot in kg

for i in range(len(T)):
    pv[i]=pow(10,10.061-(8023 / T[i]))
    gamma[i]=pv[i]/sqrt(2 * pi * m * kb * T[i])
    mdot[i]=gamma[i]*m
    m_loss[i]=mdot[i]*A*t
print(f"At {T[0]} K: mass flux = {mdot[0]:.5} kg/m^2/s; mass loss = {m_loss[0]:.5} kg")
print(f"At {T[1]} K: mass flux = {mdot[1]:.5} kg/m^2/s; mass loss = {m_loss[1]:.5} kg")
```
